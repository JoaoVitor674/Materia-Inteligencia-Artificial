**Exercicio 09: Em um array de 10 números aleatórios, normalizar valores em 0 e 1**

```python
from numpy.random import randint # Importando a função randint() do numpy para gerar números aleatórios

array = randint(0, 11, 10) # Gerando um array com 10 elementos aleatórios

minimo = min(array) # Pega o menor valor do array
maximo = max(array) # Pega o maior valor do array

normalizado = [] # Onde vai ficar os números pós normalização

for i in range(len(array)): # Repete 10 vezes pegando o tamanho do array
  normalizacao = (array[i] - minimo) / (maximo - minimo) # Faz o cálculo da normalização
  normalizado.append(float(normalizacao)) # Adiciona na lista normalizado

print(f"Essa é a lista com 10 números gerados aleatóriamente:\n{array}") # Printa o array original
print(f"Essa é a lista normalizada:\n{normalizado}") # Printa o array normalizado
```

**Exercicio 19: Calcular o angulo entre dois arrays de tamanho 20**

```python
from numpy.random import randint # Importando a função randint() do numpy para gerar números aleatórios
from numpy import dot, linalg, arccos, degrees # Importando dot() que calcula o produto escalar, linalg que contém funções de álgebra linear, arccos() que calcula o arco cosseno e degrees() que converte radianos para graus

array1 = randint(0, 21, 20) # Array de números 1
array2 = randint(0, 21, 20) # Array de números 2

produto_escalar = dot(array1, array2) # Calcula o produto escalar

norma1 = linalg.norm(array1) # Calcula a norma do primeiro array
norma2 = linalg.norm(array2) # Calcula a norma do segundo array

cosseno = produto_escalar / (norma1 * norma2) # Calcula o cosseno do ângulo

angulo = arccos(cosseno) # Calcula o ângulo em radianos
angulo_graus = degrees(angulo) # Converte o ângulo para graus

print(f"1º Array:\n{array1}\n2º Array:\n{array2}") # Printa os arrays
print(f"Ângulo entre os arrays: {angulo_graus:.2f}°") # Printa o angulo

```

Para resolver o 19º exercício foi primeiro importado a função randint() para gerar números inteiros aleatórios, dot() para calcular o produto escalar, norm() para calcular a norma, arccos() para calcular o arco cosseno e o degrees() para converter em graus. Primeiro foram criados dois arrays de comprimento 20, depois calculado o produto escalar, a seguir calculado os normas e adiante calculado o cosseno do angulo. Usando o arccos() ele calcula o angulo em radianos e depois convertido para graus com degrees. Por fim, foi printado os resultados.

**Exercicio 02: Resolver a equação diferencial dy/dx = x - y, com y(0) = 1, no intervalo de 0 a 10**

```python
from scipy.integrate import solve_ivp # Importando a função solve_ivp() do scipy para resolver equações diferenciais
from numpy import arange # Importando arange() para criar os valores de x de 0 a 10

def f(x, y): # Criando a função da equação diferencial
    return x - y # Retorna x - y, que corresponde a dy/dx

pontos = arange(0, 11, 1) # Criando os valores de x de 0 até 10

resultado = solve_ivp(f, (0, 10), [1], t_eval=pontos) # Resolvendo a equação de 0 a 10 e calculando y nos pontos definidos

for i in range(len(resultado.t)): # Percorrendo os pontos calculados
    print(f"x = {resultado.t[i]:.0f} → y = {resultado.y[0][i]:.4f}") # Printando cada valor de x e seu respectivo y
```

Para resolver o 2º exercicio de scipy foi primeiro importado a função solve_ivp() do scipy para resolver equações diferenciais e arange() para gerar os valores de x de 0 a 10. Também não sabia o que era calculo de diferencial, mas aprendi o basico: uma equação diferencial permite descobrir como os valores de y mudam conforme muda os valores de x. Nesse exercício, a equação dy/dx = x - y mostra justamente como y deve mudar conforme x muda, e y(0) = 1 indica que quando x começa em 0, y começa com o valor 1. Então foi criada uma função com a equação diferencial e depois usada a função solve_ivp() para resolver a equação no intervalo de 0 a 10, começando com y = 1. Por fim, foram gerados os valores de x nesse intervalo e mostrados os valores correspondentes de y, permitindo observar como y vai mudando conforme x aumenta.

**Exercicio 03: Encontre as raizes da função f(x) = x⁴ - 3x³ + 2**

```python
from scipy.optimize import root # Importando a função root() do scipy para encontrar raízes

def f(x): # Criando a função
    return x**4 - 3*x**3 + 2 # Retorna o valor da função

raiz1 = root(f, 1) # Procurando uma raiz começando pelo valor 1
raiz2 = root(f, 3) # Procurando uma raiz começando pelo valor 3

print(f"As raízes reais da função são aproximadamente:")
print(f"{raiz1.x[0]:.4f}") # Printando a primeira raiz
print(f"{raiz2.x[0]:.4f}") # Printando a segunda raiz
```

Para resolver o 3º exercício de scipy foi primeiro importada a função root() do scipy para encontrar as raízes da função. Nesse exercício eu já sabia o que significava encontrar a raiz de uma função, que é descobrir os valores de x que fazem a função ser igual a 0. Então foi criada uma função com f(x) = x⁴ - 3x³ + 2 e depois utilizada a função root() para encontrar as raízes reais da função. Como a função possui mais de uma raiz real, foram utilizados diferentes valores iniciais para que o scipy procurasse cada uma delas. No final, foram encontradas as duas raízes reais, que são aproximadamente 1 e 2.9196.

**Exercicio 04: Usar a função de otimização do SciPy para maximizar f(x) = -x² + 4x + 1**

```python
from scipy.optimize import minimize_scalar # Importando a função minimize_scalar() do scipy para encontrar o mínimo de uma função

def f(x): # Criando a função
    return -x**2 + 4*x + 1 # Retorna o valor da função

def negativa(x): # Criando a função negativa para encontrar o máximo de f(x)
    return -f(x) # Retorna o valor negativo de f(x)

resultado = minimize_scalar(negativa) # Procurando o menor valor da função negativa

maximo = -resultado.fun # Transformando o mínimo da função negativa no máximo da função original

print(f"O valor de x que maximiza a função é {resultado.x:.2f}") # Printando o valor de x
print(f"O valor máximo da função é {maximo:.2f}") # Printando o valor máximo
```

Para resolver o 4º exercício de scipy foi primeiro importada a função minimize_scalar() do scipy para realizar a otimização da função. **Exercicio 05: Calcular a transformada de Fourier de um sinal composto de sin(2πt) e cos(4πt)**

```python
import numpy as np # Importando o numpy para criar o tempo e as funções trigonométricas
from scipy.fft import fft, fftfreq # Importando as funções fft e fftfreq do scipy para calcular a Fourier

# Configuração do sinal de tempo
fs = 100 # Definindo a frequência de amostragem (100 pontos medidos por segundo)
duracao = 2.0 # Definindo o tempo total do sinal em segundos
N = int(fs * duracao) # Calculando o número total de pontos/amostras (100 * 2 = 200 pontos)

t = np.linspace(0.0, duracao, N, endpoint=False) # Criando o vetor de tempo com os pontos organizados
x = np.sin(2 * np.pi * 1 * t) + np.cos(2 * np.pi * 2 * t) # Criando o sinal composto pelas ondas de 1Hz e 2Hz

# Cálculo da Transformada de Fourier
X = fft(x) # Aplicando a Transformada Rápida de Fourier no sinal x
freqs = fftfreq(N, 1/fs) # Calculando a lista de frequências correspondentes para cada ponto

magnitude = np.abs(X) / N # Calculando a intensidade/volume real de cada frequência (normalizando)

# Exibição dos resultados encontrados
for f, mag in zip(freqs, magnitude): # Percorrendo cada frequência e seu respectivo valor de intensidade
    if mag > 0.1: # Filtrando para mostrar apenas as frequências que têm sinal forte (picos)
        print(f"Frequência: {f:+.1f} Hz | Amplitude: {mag:.2f}") # Printando a frequência e a amplitude encontrada
```

Para resolver o 5º exercício de SciPy, foi importado o numpy para gerar o tempo e as funções trigonométricas, e do scipy.fft foram usadas as funções fft() para a Transformada de Fourier e fftfreq() para identificar as frequências. Sinceramente, eu não sabia o que era Transformada de Fourier antes desse exercício, mas aprendi o básico: ela funciona como um "separador de ingredientes", pegando um sinal misturado e revelando quais frequências e volumes formam essa onda. No código, foi criado o sinal composto x(t) = \sin(2\pi t) + \cos(4\pi t) em um intervalo de tempo. Para resolver o 19º exercício de SciPy, foi importado as funções sin, cos e linspace do numpy, as funções fft e fftfreq do scipy.fft. Eu não conheço nada de transformada de Fourier para sinais discretos, mas aprendi que um sinal composto por senos e cossenos é formado por várias ondas sobrepostas. A função da FFT é para separar essas ondas para mostrar quais frequências formam o sinal. No código, foi criado um sinal com 2 frequências definidas (5Hz e 20Hz) em um intervalo de tempo de 1 segundo e foi aplicado a fft(). **Exercicio 20: Resolver um sistema de equações lineares com 3 variaveis**

```python
from numpy import array # Importando a função array() do numpy para estruturar a matriz e o vetor
from scipy.linalg import solve # Importando a função solve() do scipy.linalg para resolver o sistema linear

# Criando a matriz A de coeficientes das 3 variáveis (x, y, z)
A = array([
    [2,  1, -1],
    [-3, -1,  2],
    [-2,  1,  2]
])

# Criando o vetor B dos resultados das equações
B = array([8, -11, -3])

# Calculando a solução do sistema A * X = B usando a função solve() do scipy
solucao = solve(A, B)

# Separando e printando os resultados encontrados para x, y e z
x, y, z = solucao

print(f"Matriz dos coeficientes (A):\n{A}") # Printando a matriz A
print(f"\nVetor dos termos independentes (B):{B}") # Printando o vetor B
print(f"\nSolução encontrada: x = {x:.2f}, y = {y:.2f}, z = {z:.2f}") # Printando a solução final
```

Para resolver o 20º exercício de SciPy, foi importado a função array() do numpy para organizar a estrutura dos dados e a função solve() do scipy.linalg para calcular a solução. Resolver um sistema de 3 variáveis (x, y e z) significa encontrar os valores que resolvem três equações lineares ao mesmo tempo. No código, foi separado o sistema na matriz de coeficientes A e no vetor de resultados B. Em seguida, foi passado ambos para a função solve(), e o SciPy realizou as operações matriciais, encontrando a solução exata para as três variáveis do sistema.

**Exercicio 08: Plotar um gráfico de linha com intervalos de confiança (sombramento) para a função y = sin(x)**

```python
import numpy as np # Importando a biblioteca numpy
import matplotlib.pyplot as plt # Importando a biblioteca matplotlib

x = np.linspace(0, 2 * np.pi, 100) # Criando 100 valores de x entre 0 e 2π

y = np.sin(x) # Calculando os valores de y = sin(x)

erro = 0.2 # Definindo o tamanho do intervalo de confiança

limite_superior = y + erro # Calculando o limite superior do intervalo
limite_inferior = y - erro # Calculando o limite inferior do intervalo

plt.plot(x, y) # Criando o gráfico de linha da função seno

plt.fill_between( # Criando o sombreamento entre os limites do intervalo
    x,
    limite_inferior, # Limite inferior do sombreamento
    limite_superior, # Limite superior do sombreamento
    alpha=0.2 # Transparencia do sombreamento
)

plt.title("Função y = sin(x)") # Adicionando título
plt.xlabel("X") # Definindo o nome do eixo x
plt.ylabel("Y") # Definindo o nome do eixo y

plt.show() # Exibindo o gráfico
```

Para resolver o 8º exercício foi usado o linspace() para criar os valores de x, sin() para calcular os valores da função, plot() para criar o gráfico de linha, fill_between() para criar o sombreamento, title() para colocar um título, xlabel() e ylabel() para os nomes dos eixos e show() para exibir o gráfico, usando numpy e matplotlib. Para começar foram criados 100 valores de x entre 0 e 2π (para dar uma ciclo completo do seno), depois calculados os valores de y = sin(x) e definido um erro de 0.2 para representar o intervalo. Em seguida, foram calculados os limites superior e inferior do intervalo e criado o gráfico de linha da função seno. Depois, foi adicionado o sombreamento entre os dois limites. Por fim, foram definidos o título e os nomes dos eixos e o gráfico foi exibido.

```python
import matplotlib.pyplot as plt # Importando a biblioteca matplotlib

categorias = ["A", "B", "C", "D", "E"] # Definindo as categorias

valores = [20, -15, 30, -10, 25] # Definindo valores positivos e negativos

cores = [] # Criando uma lista vazia para armazenar as cores

for valor in valores: # Percorrendo cada valor
    if valor >= 0: # Verificando se o valor é positivo ou zero
        cores.append("blue") # Adicionando azul para valores positivos
    else:
        cores.append("red") # Adicionando vermelho para valores negativos

plt.bar(categorias, valores, color=cores) # Criando o gráfico de barras com as cores definidas

plt.title("Valores positivos e negativos") # Adicionando título
plt.xlabel("Categorias") # Definindo o nome do eixo x
plt.ylabel("Valores") # Definindo o nome do eixo y

plt.axhline(0, color="black", linewidth=1) # Adicionando uma linha horizontal no zero

plt.show() # Exibindo o gráfico
```

Para resolver o 18º exercício foi usado o bar() para criar o gráfico de barras, append() para por as cores em uma lista, title() para por um título, xlabel() e ylabel() para definir os nomes dos eixos, axhline() para por uma linha horizontal no zero e show() para exibir o gráfico, utilizando matplotlib. Para começar foram definidas cinco categorias e valores positivos e negativos. Em seguida, cada valor foi analisado e valores positivos receberam a cor azul e valores negativos receberam a cor vermelha. Depois, foi criado o gráfico de barras utilizando as categorias, os valores e as cores definidas. Por fim, foi adicionada uma linha horizontal no zero, o título e os nomes dos eixos, e o gráfico foi exibido.

**Exercicio 19: Plotar um gráfico de linha com duas séries e adicione uma legenda para diferenciá-las**

```python
import numpy as np # Importando a biblioteca numpy
import matplotlib.pyplot as plt # Importando a biblioteca matplotlib

x = np.arange(0, 11) # Criando os valores de x de 0 a 10

y1 = x # Definindo a primeira série
y2 = x**2 # Definindo a segunda série

plt.plot(x, y1, label="y = x") # Criando a primeira linha com a legenda
plt.plot(x, y2, label="y = x²") # Criando a segunda linha com a legenda

plt.title("Duas séries") # Adicionando título
plt.xlabel("X") # Definindo o nome do eixo x
plt.ylabel("Y") # Definindo o nome do eixo y

plt.legend() # Adicionando a legenda

plt.show() # Exibindo o gráfico
```

Para resolver o 19º exercício foi usado o arange() para os valores de x, plot() para criar o gráfico de linha, legend() para adicionar uma legenda, title() para adicionar um título, xlabel() e ylabel() para definir os nomes dos eixos e show() para exibir o gráfico, utilizando numpy e matplotlib. Para começar foi definido os valores de x de 0 a 10 e as séries, y1 = x e y2 = x². Em seguida, foram criados os dois gráficos de linha e definidos nomes para cada série usando o parâmetro label. Depois, foi adicionada a legenda para diferenciar as duas séries. Por fim, foi adicionado o título, definidos os nomes dos eixos e exibido o gráfico.

**Exercicio 20: Adicionar uma grade com espaçamento personalizado a um gráfico de linha para y = x²**

```python
import numpy as np # Importando a biblioteca numpy
import matplotlib.pyplot as plt # Importando a biblioteca matplotlib

x = np.arange(-10, 11) # Criando os valores de x de -10 a 10

y = x**2 # Calculando os valores de y = x²

plt.plot(x, y) # Criando o gráfico de linha

plt.title("Função y = x²") # Adicionando título
plt.xlabel("X") # Definindo o nome do eixo x
plt.ylabel("Y") # Definindo o nome do eixo y

plt.xticks(np.arange(-10, 11, 2)) # Definindo o espaçamento dos valores do eixo x
plt.yticks(np.arange(0, 101, 20)) # Definindo o espaçamento dos valores do eixo y

plt.grid(True, linestyle="--") # Adicionando a grade

plt.show() # Exibindo o gráfico
```

Para resolver o 20º exercício foi usado o arange() para criar os valores de x, plot() para criar o gráfico de linha, xticks() e yticks() para definir o espaçamento dos valores dos eixos, grid() para adicionar a grade, title() para adicionar um título, xlabel() e ylabel() para definir os nomes dos eixos e show() para exibir o gráfico, utilizando numpy e matplotlib. Para começar foram criados os valores de x de -10 a 10 e calculados os valores da função y = x². Em seguida, foi criado o gráfico de linha e definido o espaçamento dos valores dos eixos x e y, fazendo com que a grade também acompanhasse esse espaçamento. Por fim, foi adicionada a grade, o título, os nomes dos eixos e exibido o gráfico.
