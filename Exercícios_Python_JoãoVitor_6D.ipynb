#Exercicio 12: Contagem de caracteres

```python
Palavra = input("Me fale uma palavra: ") # Pede uma palavra

contador = dict() # Dicionario para contar quantas vezes repete cada palavra

for i in range(len(Palavra)): # Para repetir a quantidade de vezes equivalente a quantidade de letras da palavra
  if Palavra[i] in contador: # Se a letra se tiver no dicionario ele soma, senão ele começa com 1
    contador[Palavra[i].lower()] += 1
  else:
    contador[Palavra[i].lower()] = 1

for chave, valor in contador.items(): # Printa a chave e o valor que está no dicionario
  print(f"Letra {chave.upper()} aparece {valor} vezes")
```

Usa-se pela primeira vez o dicionario e o conceito de que toda String é uma lista de caracteres. No exercício é feito primeiro um For que se repete a quantidade de vezes corespondente ao tamanho da String, logo após é feito uma verificação letra por letra, onde se essa letra já tiver no dicionario ele soma 1 no valor daquela chave, e se não tiver ainda no dicionario ele recebe o valor 1. No final é usado um For para mostrar todas as chaves e valores do dicionario.

#Exercicio 20: Matriz Transposta

```python
matriz = [
    [4, 5, 6],
    [5, 8, 11]
]

linhas = len(matriz) # Pega quantas linhas tem
colunas = len(matriz[0]) # Pega quantas colunas tem

transposta = [] # Onde vai ficar a transposta

for i in range(colunas): # Repete pela quantidade de colunas
  linha = []

  for j in range(linhas): # Percorre as linhas para pegar os valores das colunas atuais
    linha.append(matriz[j][i]) # Nova linha com os valores da coluna atual

  transposta.append(linha) # Adiciona a nova linha na transposta

for i in transposta:
  print(i) # Printa a matriz transposta
```

Usa-se len() para identificar a quantidade de linhas e colunas da matriz atual. Logo em seguida, é usado for para percorrer as colunas e linhas, adicionando os valores das colunas em novas linhas para formar a matriz transposta.

#Exercicio 21: Busca Binária

```python
numeros = [1, 3, 5, 8, 9, 25, 27, 32, 49] # Lista ordenada

numero_procurado = 32 # Variavel para colocar o número para fazer a busca binaria

inicio = 0 # Primeira posição da lista
fim = len(numeros) - 1 # Ultima posição da lista

print(f"Tem com base a lista de baixo, usaremos o método da busca binaria para encontrar a posição do número {numero_procurado} nele") # Mostra a lista de números
print(numeros)

while inicio <= fim:
  metade = (inicio + fim) // 2 # Posição da metade da lista

  if numeros[metade] == numero_procurado: # vê se o numero na metade é igual ao numero procurado
    print(f"O número {numero_procurado} foi encontrado na posição: {metade + 1}")
    break
  elif numeros[metade] < numero_procurado: # Vai aumentando de 1 em 1 até achar em qual parte ele tá a partir da metade
    inicio = metade + 1
  else:
    fim = metade - 1 # Ou diminuindo de 1 em 1 se for maior que a metade
```

Usa-se o método de busca binária em uma lista ordenada, onde primeiro é definido a primeira posição (0) e a última posição da lista (len(numeros) - 1). Depois, usando o while, o código pega o número que está no meio e verifica se é o número procurado. Se não for, ele vai somando ou subtraindo de 1 em 1 até encontrar o número.

#Exercicio 22: Criptografia de Cifra de Cesar

```python
def criptografar(frase, contador): # Função para criptografar e pede a palavra e quantas casas vai avançar
  palavra_criptografada = [] # Onde vai ser adicionada a palavra criptografada

  for i in frase.lower(): # Percorre cada letra da palavra fornecida e usa só a palavra em minusculo
    num = int(ord(i) - ord("a") + contador) % 26 # Adiciona a quantidade que se quer deslocar e o % 26 faz o alfabeto voltar para o final quando necessário
    letra = chr(num + ord("a")) # Retorna o valor ASCII dela
    palavra_criptografada.append(letra) # Vai adicionando as letras criptografadas na lista

  return "".join(palavra_criptografada) # Junta a frase e a retorna no final como uma String só


crip1 = criptografar("Fabiano", 2) # Usando a função com avanço de duas casas
print(crip1.capitalize()) # Printa com a primeira letra em maiuscula
```

Foi criado a função criptografar() com os atributos frase e contador, na qual são qual frase vai ser criptografada e quantas casas vão ser avançadas. O programa pega somente a frase com letra minuscula e se não tiver ele transforma, pega o valor ASCII para calcular com o contador e depois retorna para letra com o chr() já criptografado. Ele também faz validação se for tipo a letra Z que se somar 2, por exemplo tem que virar B e não bugar.

#Exercicio 24: Validação de CPF

```python
def verificadorCPF(cpf):
  # Verifica se o CPF tem 11 dígitos e se são números validos
  if len(cpf) != 11 or not cpf.isdigit():
      print("O CPF digitado é inválido!")
  else:
      # Primeiro dígito verificador
      soma = 0

      for i in range(9): # Faz a soma dos digitos multiplicados numa ordem inversa (1º digito * 10, 2º digito * 9)
          soma += int(cpf[i]) * (10 - i)

      resto = soma % 11 # Pega o resto da divisão por 11

      if resto < 2: # Vê se o 1º digito é zero for menor que 2, mas se for maior subtrai 11 do resto e defini o 1º digito
          primeiro_digito = 0
      else:
          primeiro_digito = 11 - resto

      # Segundo dígito verificador
      soma = 0

      for i in range(10): # Faz a soma dos digitos multiplicados numa ordem inversa, mas agora contando o 1º digito verificador (1º digito * 11, 2º digito * 10)
          soma += int(cpf[i]) * (11 - i)

      resto = soma % 11 # Pega o resto da divisão por 11

      if resto < 2: # Vê se o 2º digito é zero for menor que 2, mas se for maior subtrai 11 do resto e defini o 2º digito
          segundo_digito = 0
      else:
          segundo_digito = 11 - resto

      # Compara os dígitos calculados com os do CPF fornecido
      if primeiro_digito == int(cpf[9]) and segundo_digito == int(cpf[10]):
          print(f"O CPF {cpf} é válido!")
      else:
          print(f"CPF inválido! Para ser valido os digitos finais tinham que ser {primeiro_digito}{segundo_digito}") # Imprime quais deveriam ser os digitos finais caso o CPF seja invalido

verificadorCPF("52998224725") # Usando a função

```

O programa tem uma função para verificar se os dois ultimos digitos do cpf são validos. Para isso, primeiro ele analisa se foi digitado um numero de 11 digitos e com digitos. Depois ele começa as contas do primeiro digito final, que é fazendo a soma dos nove primeiros digitos multiplicados em ordem decrescente de 9 à 2, depois pegando o resto da divisão por 11 dessa soma e se for menor que dois o 1º digito é 0 senão é 11 - resto. O segundo digito final é semelhante mas agora faz a conta com o 1º digito final descoberto. No final ele imprime se é valido ou se não for ele mostra quais deveriam ser os números corretos pra ser validos.

#Exercicio 29: Números FizzBuzz

```python
fizz = [] # Onde será recebido os números classificados como Fizz
buzz = [] # Onde será recebido os números classificados como Buzz
fizzbuzz = [] # Onde será recebido os números classificados como FizzBuzz

# Nesse for é classificado os numeros de 1 à 100 em Fizz, Buzz e FizzBuzz
for i in range(1,101):
  if i % 3 == 0 and i % 5 == 0:
    fizzbuzz.append(i)
  elif i % 3 == 0:
    fizz.append(i)
  elif i % 5 == 0:
    buzz.append(i)

print("Os números de 1 à 100 são classificados: \nFizz:")

for i in fizz: # Printa os Fizz
  print(i, end=" ")

print("\nBuzz: ")

for i in buzz: # Printa os Buzz
  print(i, end=" ")

print("\nFizzBuzz: ")

for i in fizzbuzz: # Printa os FizzBuzz
  print(i, end=" ")
```

```python
palavra = "carro" # Palavra definida
vidas = 6 # Quantidade de vidas
tentativas = "" # Onde ficará armazenadas as letras que forem sendo acertadas

print("---- Jogo da Forca ----")
print("Você tem 6 vidas! A palavra tem 5 letras")

while vidas > 0: # Enquanto o numero de vidas for maior que zero
  tentativa = input("Dê seu palpite digitando uma letra: ")
  if tentativa in palavra: # Verifica se a letra da tentativa está na palavra
    tentativas += tentativa # Armazena a letra se acertar

    for letra in palavra: # Transpassa letra por letra da palavra
      if letra in tentativas: # Se ela estiver onde fica armazenada as que já foram acertadas, ela printa a letra
        print(letra, end="")
      else: # Senão só printa _  , que é a letra que ainda falta
        print("_ ",end="")
    print()

    if all(letra in tentativas for letra in palavra): # Verifica se todas as letras já foram acertadas e dá a vitoria se sim
            print(f"Você venceu! A palavra era {palavra}. Parabens!!")
            break
  else: # Caso a letra não esteja na palavra ele tira uma vida
    vidas -= 1 # - uma vida
    print(f"Errou! Perdeu uma vida! Agora você tem {vidas} vidas")
else: # Quando perde todas as vidas
  print(f"Você perdeu! Errou as 6 vezes possiveis. A palavra era {palavra}")
```

# Exercicio 32: Jogo da Velha

```python
def mostrar_tabuleiro(): # Criando uma função para exibir o tabuleiro durante a partida
    print(f"{tabuleiro[0]} | {tabuleiro[1]} | {tabuleiro[2]}")
    print("--+---+--")
    print(f"{tabuleiro[3]} | {tabuleiro[4]} | {tabuleiro[5]}")
    print("--+---+--")
    print(f"{tabuleiro[6]} | {tabuleiro[7]} | {tabuleiro[8]}")

tabuleiro = [" ", " ", " ", # Tabuleiro onde vai ser jogado
             " ", " ", " ",
             " ", " ", " "]

vitorias = [ # Contem todas as sequencias vencedoras
    [0, 1, 2],
    [3, 4, 5],
    [6, 7, 8],
    [0, 3, 6],
    [1, 4, 7],
    [2, 5, 8],
    [0, 4, 8],
    [2, 4, 6]]

jogador = "X" # Começa com o jogador X e vai trocando pelo O durante o jogo

while True:
    mostrar_tabuleiro() # Mostra o tabuleiro
    posicao = int(input(f"Jogador {jogador}, escolha uma posição: ")) # Pedi uma jogada

    if posicao < 0 or posicao > 8: # Verifica se o número escolhido é menor ou maior do que os permitidos no jogo
      print("Posição inválida!")
      continue # Retorna pro começo do while

    if tabuleiro[posicao] != " ": # Não deixa subscrever uma posição já ocupada
      print("Essa posição já está ocupada!")
      continue # Retorna pro começo do while

    tabuleiro[posicao] = jogador # Adiciona uma jogada

    venceu = False # Para verificar se venceu

    for combinacao in vitorias: # Vê se o jogador venceu
      if tabuleiro[combinacao[0]] == tabuleiro[combinacao[1]] == tabuleiro[combinacao[2]] != " ":
        print(f"Parabens! Jogador {jogador} venceu!")
        mostrar_tabuleiro() # Mostra o tabuleiro final!
        venceu = True # Retorna true se vencer e faz quebrar o while
        break # Quebra esse if

    if venceu: # Agora ele quebra o while e o jogo acaba
      break

    if " " not in tabuleiro: # Verifica empates
      print("Jogo empatou! Ninguem venceu")
      break

    if jogador == "X": # Faz a troca dos jogadores
      jogador = "O"
    else:
      jogador = "X"
```

O programa começa criando uma função para mostrar o tabuleiro a cada jogada, o proprio tabuleiro do jogo e uma lista com todas as sequencias vitoriosas. O jogo começa com o jogador X e mostrando o tabuleiro vazio, e começa pedindo uma jogada do jogador X, onde ele só pode escolher numeros de 0 a 8 (verifica se não for) e um numero que ainda não foi jogado. Quando dá alguma combinação vitoriosa ele para o while e dá a vitoria para o jogador vitorioso, e se nenhuma posição der vitoria, é dado o empate.

# Exercicio 37: Funções Recursivas

```python
def fatorial(n): # Criando uma função que vai chamar ela mesma (Função Recursiva) para calcular fatorial
    if n == 1: # Criando a parada do método
        return 1

    return n * fatorial(n - 1) # Ele chama ele proprio para calcular ele vezes ele menos - 1 até chegar a 1

fatorial(6) # Chamando o método
```

A função gerada é uma função recursiva, pois ela chama ela mesma, e assim ela calcula o fatorial, sempre multipliccando ela por ela menos 1 até chegar em 1 e ela parar de chamar ela mesma

# Exercicio 46: Calcular Média e Desvio Padrão

```python
from math import sqrt # Importando do math o método para calcular raiz quadrada

numeros = [5, 2, 6, 8, 7] # Lista de números
soma = 0 # Onde vai ser somado todos os números da lista acima para calcular a média

for i in numeros: # For que percorre toda os números da lista
  soma += i # Adiciona/soma os números um por um da lista na variavel soma

media = soma/len(numeros) # Cálculo da média dividindo a soma pelo tamanho da lista

v = [] # Onde vai receber a parte de cima do calculo de variancia

for i in numeros: # Percorre todos os números da lista
  a = (i - media) ** 2 # Parte de cima do calculo de variancia
  v.append(a) # Adiciona no v

somav = 0 # Onde vai receber a soma dos quadrados realizados acima

for i in v: # Percorre todo o v e os soma
  somav += i

variancia = somav/len(numeros) # Agora faz o calculo completo da variancia
desvio_padrao = sqrt(variancia) # E por fim se tira a raiz quadrada da variancia para ter o desvio padrão

print(f"Essa é a média: {media:.2f}") # Print da média

print(f"Esse é o desvio padrão: {desvio_padrao:.2f}") # Print do desvio padrão
```

O programa começa importando da biblioteca math o método sqrt para calcular raiz quadrada, já que o desvio padrão é a raiz quadrada da variancia. Dada uma lista de numeros, primeiro é calculado a média realizando a soma de todos os números da lista e depois dividindo pelo tamanho (len) da lista. Tendo a média calculada, é possivel fazer a primeira parte do calculo da variancia que é o calculo de cada numero da lista menos a média, tudo isso elevado ao quadrado e armazenando na lista v. Depois é realizado a soma de todos esses numeros da lista v na variavel somav. Logo após, é pego o somav e dividido pelo tamanho (len) da lista numeros original, assim extraindo a variancia. E por fim, se calcula o desvio padrão tirando a raiz quadrada com sqrt da variancia. Depois são todos printados na tela.

# Exercicio 48: Ordenação Por Inserção

```python
numeros = [5, 2, 6, 8, 7, 4, 1, 3] # Lista de números

for i in range(1, len(numeros)): # Percorre a lista começando pelo 2º elemento
    numatual = numeros[i] # Guarda o número atual
    num = i - 1 # Pega a posição do número anterior ao atual

    while num >= 0 and numeros[num] > numatual: # Repete enquanto houver elementos à esquerda maiores que o atual
        numeros[num + 1] = numeros[num] # Move o número maior uma posição para a direita
        num -= 1 # Volta uma posição para continuar comparando

    numeros[num + 1] = numatual # Coloca o número atual na posição correta

print(numeros) # Printa os números ordenados por inserção
```

O programa começa definindo uma lista de números desordenada. Depois ele vai comparando os némeros e colocando eles na ordem correta, depois printa eles já ordenados.

```python
Palavra = input("Me fale uma palavra: ") # Pede uma palavra

contador = dict() # Dicionario para contar quantas vezes repete cada letra

for i in range(len(Palavra)): # Para repetir a quantidade de vezes equivalente a quantidade de letras da palavra
  if Palavra[i] in contador: # Se a letra se tiver no dicionario ele soma, senão ele começa com 1
    contador[Palavra[i].lower()] += 1
  else:
    contador[Palavra[i].lower()] = 1

for chave, valor in contador.items(): # Printa a chave e o valor que está no dicionario
  print(f"Letra {chave.upper()} aparece {valor} vezes")
```
